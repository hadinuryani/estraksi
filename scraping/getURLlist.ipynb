```python
# Import library yang dibutuhkan
import undetected_chromedriver as uc  # Pengendali browser Chrome & bypass proteksi Cloudflare WAF
from bs4 import BeautifulSoup           # Parsing dan pencarian elemen dalam struktur HTML
import time                             # Jeda waktu (delay) saat memuat halaman web
import os                               # Operasi file dan path sistem
import warnings                         # Menyaring peringatan yang tidak krusial
```

```python
# Inisialisasi Chrome driver 
# Browser dibuka satu kali di awal agar hemat memori
driver = uc.Chrome(version_main=153)
```

```python
def getURLfromWeb(url):
    print(f"Mengakses: {url}")
    driver.get(url)
    
    # Jeda 5 detik agar Cloudflare challenge selesai diverifikasi dan halaman termuat penuh
    time.sleep(5)

    # Ambil source HTML halaman yang sudah selesai dirender oleh browser
    html = driver.page_source
    soup = BeautifulSoup(html, 'html.parser')
    
    # Cari semua tag anchor <a> yang memiliki tautan href
    result1 = soup.find_all('a', href=True)
    
    urlHasil = []
    for eachResult1 in result1:    
        href = eachResult1['href']
        # Filter: Hanya ambil link yang menuju ke halaman detail putusan Mahkamah Agung
        if 'https://putusan3.mahkamahagung.go.id/direktori/putusan/' in href:
            if href not in urlHasil:
                urlHasil.append(href)

    print(f"Ditemukan {len(urlHasil)} link putusan\n")
    return urlHasil
```

```python
def main():
    warnings.filterwarnings('ignore')
    
    # URL awal (halaman 1) dan pola URL untuk halaman selanjutnya (page/2, page/3, dst.)
    url1 = 'https://putusan3.mahkamahagung.go.id/direktori/index/pengadilan/pn-blitar/kategori/pidana-umum-1.html'
    url2 = 'https://putusan3.mahkamahagung.go.id/direktori/index/pengadilan/pn-blitar/kategori/pidana-umum-1/page/'
    
    listHasil = []
    ulang = 10  # Mengambil 10 halaman daftar putusan
    
    startTime = time.time()
    
    # Buka file hasilListURLPage.txt untuk menyimpan kumpulan link putusan
    file_hasil = open("hasilListURLPage.txt", "w", encoding='UTF8')
    
    for i in range(ulang):
        if i == 0:
            url = url1
            print("=== Halaman 1 ===")
        else:            
            n = i + 1
            url = url2 + str(n) + '.html'
            print(f"=== Halaman {n} ===")

        # Ambil link putusan dari halaman aktif
        listHasil = getURLfromWeb(url)

        # Tulis link ke file teks
        for listURL in listHasil:
            file_hasil.write(listURL + "\n")

    file_hasil.close()
    endTime = time.time()
    
    print('Semua halaman berhasil diambil!')
    print('Total waktu proses : ', round(endTime - startTime, 2), ' Detik')
        
main()
```

```python
# Tutup sesi browser setelah proses pengambilan URL selesai
driver.quit()
```
